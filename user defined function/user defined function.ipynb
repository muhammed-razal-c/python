```python
def monthly_sales_analysis(previous_sales, current_sales):
    difference = current_sales - previous_sales
    if previous_sales == 0:
        if current_sales == 0:
            percentage_change = 0
        else:
            percentage_change = None 
    else:
        percentage_change = (difference / previous_sales) * 100

    if difference > 0:
        status = "Sales increased"
    elif difference < 0:
        status = "Sales decreased"
    else:
        status = "Sales remained unchanged"


   
    print(f"Difference     : {difference}")
    print(f"Status         : {status}")

    if percentage_change is None:
        print("Percentage Change: Undefined (previous sales were 0)")
    else:
        print(f"Percentage Change: {abs(percentage_change):.2f}%")

monthly_sales_analysis(50000, 40000)
```

```python
def inventory_analysis (product_name, opening_stock, units_sold, reorder_level):
    remaining_stock=opening_stock-units_sold 
    opening_stock_sold= (units_sold/opening_stock)*100
    print ("Product name: ",product_name)
    print("Opening_Stock:",opening_stock)
    print("Units sold: ",units_sold)
    print("Reorder level: ", reorder_level)

    if remaining_stock>reorder_level:
        print("Stock status: High")
    elif remaining_stock==reorder_level:
        print ( "Stock status: Optimal")
    elif remaining_stock<reorder_level:
        print ( "Stock status: Low")
    else:
        print ("Product Unavailable")
inventory_analysis ("Pen", 100,50,50)
```

```python
def sales_trend( month1_sales,month2_sales,month3_sales):
    total_sales=month1_sales+month2_sales+month3_sales
    Average_sales=round(total_sales/3,2)
    percentage_1=(month2_sales-month1_sales)/month1_sales*100
    percentage_2=(month3_sales-month2_sales)/month2_sales*100
    print("Total Sales   :",total_sales)
    print("Average Sales :",Average_sales)
    print("Month 1 To Month 2 Growth :",percentage_1,"%")
    print("Month 2 To Month 3 Growth :",percentage_2,"%")

    if month2_sales > month3_sales:
        print("Status : Sales decreased")
    elif month2_sales < month3_sales:
        print("Status : Sales increased")
    else:
        print("Status : unchanged")
```

```python
sales_trend(50000,60000,75000)
```

```python
def sales_analysis(sales):
    total_sales = sum(sales)
    average_sales = total_sales/len(sales)
    print("total sales   :",total_sales)
    print("average sales :",average_sales)
    for i in range(1,len(sales)):
        previous = sales[i-1]
        current = sales[i]
        difference = current-previous
        percentage = (difference/previous)*100
        print("difference    :",difference)
        print("percentage    :",percentage,"%")
        if current > previous:
            print("Status        : Sales increased")
        elif current < previous:
            print("Status        : Sales decreased")
        else:
            print("Status        : Sales unchanged")
        print("-------------------------")
        
        
    
```

```python
sales_analysis([1000,1200,1400,1400,1600,1500,1300,1600])
```

```python
def daily_expense_analysis(daily_sales):
    for i in range(1,len(daily_sales)):
        previous = daily_sales[i-1]
        current = daily_sales[i]
        comparison = current - previous
        print("comparison :",comparison)
        if current > previous:
            print("Status     : Sales increased")
        elif current < previous:
            print("Status     : Sales decreased")
        else:
            print("Status     : Sales unchanged")
        print("___________________________________________________")
```

```python
daily_expense_analysis([2000,2100,1900,2500,2300,2300,1800,2000])
```

```python
def sales_target_achievement_analysis(monthly_sales,sales_target):
    total_sale_of_month = sum(monthly_sales)
    print("Total sales of months    :",total_sale_of_month)
    average_monthly_sales = total_sale_of_month/len(monthly_sales)
    print("Average sales of monthS  :",average_monthly_sales)
    count = 0 
    for s in monthly_sales:
        if s >= sales_target:
            count = count + 1
    print("months achieved target   :",count)
    percentage = (count/len(monthly_sales))*100
    print("percentage changed       :",percentage)
    if average_monthly_sales > sales_target:
        print("Average sales met the monthly target")
    else:
        print("Average sales did not met the monthly target")
```

```python
monthly_sales=[22000,20000,21000,23000,25000,24000,24000,23000,21000]
sales_target_achievement_analysis(monthly_sales,22500)
```
