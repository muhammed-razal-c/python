```python
def monthly_sales(sales,sales_target):
    total_sales = sum(sales)
    print("Total Sales             :",total_sales)
    average_monthly_sales = (total_sales/len(sales))*100
    print("Average Sales           :",average_monthly_sales)
    count = 0
    for i in sales:
        if i >= sales_target:
            count = count +1
    print("Months Target Achieved  :" , count)
    if average_monthly_sales > sales_target:
        print("Overall performance     : target Achieved")
    else:
        print("Overall performance     : target not Achieved")
        
sales = [8000, 12000, 9000, 15000, 11000]   
```

```python
monthly_sales(sales,10000)
```

```python
def Employee_Attendance_Analysis(Attendance):
    present = Attendance.count(1)
    absent  = Attendance.count(0)
    percentage = (present/len(Attendance))*100
    print("Days present   :",present)
    print("Days absent    :",absent)
    print("Attendance     :",percentage)
    if percentage >= 80:
        print("Status         : Good Attendance")
    else:
        print("Status         : Bad Attendance")
```

```python
Employee_Attendance_Analysis([1,1,0,1,1])
```

```python
def Product_Sales_Analysis(units):
    total_units = sum(units)
    average_units = total_units/len(units)
    highest = max(units)
    count = 0 
    for u in units:
        if u >= 20:
            count = count + 1
    print("Total Units Sold      :",total_units)
    print("Average Daily Sales   :",average_units)
    print("Highest Daily Sales   :",highest)
    print("Days Above 20 Units   :",count)
```

```python
Product_Sales_Analysis([15, 25, 30, 10, 35])
```

```python
def Customer_Rating_Analysis(ratings):
    total = sum(ratings)
    average = round(total/len(ratings),2)
    count = 0
    for r in ratings:
        if r >= 4:
            count = count + 1
    low = 0
    for l in ratings:
        if l < 3 :
            low = low + 1
    print("Average Rating   :",average)
    print("Positive Rating  :",count)
    print("Low Rating       :",low)
    if average >= 4:
        print("Status           : Good Customer Satisfaction")
    else:
        print("Status           : Needs Improvement")
```

```python
Customer_Rating_Analysis([5, 4, 3, 2, 5, 4])
```

```python
def Daily_Expense_Analysis(expense):
    total=sum(expense)
    average=round(total/len(expense),2)
    highest=max(expense)
    count = 0 
    for e in expense:
        if e >=1000:
            count = count +1
    print("Total Expenses          :",total)
    print("Average Daily Expenses  :",average)
    print("Highest Expense         :",highest)
    print("Days Above 1000         :",count)
    if average <=1000:
        print("Status                  : Within Budget")
    else:
        print("Status                  : Over Budget")
expense = [500, 1200, 800, 1500, 900, 700, 1600]
```

```python
Daily_Expense_Analysis(expense)
```
