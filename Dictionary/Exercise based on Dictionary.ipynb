## Set C -- Challenge

```python
# 1.
record = {}

while True:
    key = input("Enter key (or 'stop'): ")
    if key == "stop":
        break
    value = input("Enter value: ")
    record.update({key: value})

print(record)
```

```python
# 2.
sentence = "to be or not to be"
counts = {}

for word in sentence.split():
    counts.update({word: counts.get(word, 0) + 1})

print(counts)
```

```python
# 3.
marks = {"Asha": 35, "Ravi": 78, "Meena": 29, "John": 60}

for name in list(marks.keys()):
    if marks[name] < 40:
        marks.pop(name)

print(marks)
```
