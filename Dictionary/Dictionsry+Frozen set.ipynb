#### 1. add()

```python
fruits = {"Apple","Orange"}
fruits.add("Mango")
print(fruits)
```

#### 2.update()

```python
fruits = {"Apple"}
fruits.update(["Orange","Mango"])
print(fruits)
```

#### 3.remove()

```python
fruits = {"Apple","Orange","Mango"}
fruits.remove("Orange")
print(fruits)
```

#### 4.discard()

```python
fruits = {"Apple","Orange"}
fruits.discard("Orange")
print(fruits)
```

#### 5.pop()

```python
fruits = {"Apple","Orange","Mango"}
fruits.pop()
print(fruits)
```

#### 6.clear()

```python
fruits = {"Apple","Orange"}
fruits.clear()
print(fruits)
```

#### 7.copy()

```python
fruits = {"Apple","Orange"}
new_set = fruits.copy()
print(new_set)
```

#### 8.union()

```python
Fruits = {"Apple","orange","mango"}
impfruits = {"mango","kiwi","avacado"}

print(Fruits.union(impfruits))
```

#### 9.intersection()

```python
A = {1,2,3}
B = {2,3,4}

print(A.intersection(B))
```

#### 10.difference()

```python
A = {1,2,3}
B = {2,3,4}

print(A.difference(B))
```

#### 11.symmetric_difference()

```python
A = {1,2,3}
B = {2,3,4}

print(A.symmetric_difference(B))
```

# Frozen set

#### A Frozenset is an immutable version of a set. Once created, its elements cannot be added, removed, or changed.

###### ->Unordered
###### ->Immutable
###### ->Unique elements only
###### ->Supports set operations like union(), intersection(), and difference()

```python
# Syntax
frozenset_name = frozenset(iterable)
```

```python
numbers=frozenset({1,2,3,4,5,6,7,8,9,0})
print(numbers)
```
