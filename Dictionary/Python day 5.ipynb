# Part 1: Dictionary

### A Dictionary is an unordered , mutable collection of key-value pairs.
###### -Stores data as key : value
###### -Keys must be unique
###### -Values can be duplicated
###### -Mutable (can be modified)
###### -Fast lookup using keys


```python
#Example
student = {
    "Name": "John",
    "Age": 22,
    "Course": "Data Science",
    "Marks": 95
}

print(student)
```

```python
print(student["Name"])
print(student["Age"])
```

```python
Sales ={"Sales ID":["S001","S002","S003","S004","S005"],
        "Product ID":["P001","P002","P003","P004","P005"],
        "Product Name":["Laptop","Mobile","Headset","computer","UPS"],
        "Price":[68990,42000,1200,34000,56000]}
Order ={"Order ID":["O001","O002","O003","O004","O005"],
        "Order Name":["Laptop","Mobile","Headset","computer","UPS"]}

print(Sales,Order)
```

```python
print(Sales["Sales ID"])
print(Sales["Product ID"])
print(Sales["Product Name"])
print(Sales["Price"])
print(Order["Order ID"])
print(Order["Order Name"])
```

## Dictionary methods

#### 1.get()
###### Returns the value of a key.

```python
# Example
student = {"Name":"Najil","Age":"21","Course":"Data"}
print(student.get("Name"))
```

#### 2.keys
###### Returns all keys.

```python
print(student.keys())
print(Sales.keys())
print(Order.keys())
```

#### 3.values()
###### Returns all values.

```python
print(Sales.values())
print(student.values())
print(Order.values())
```

#### 4.item()

```python
print(Sales.items())
print(student.items())
print(Order.items())
```

#### 5.update()
###### Updates the dictionary.

```python
student.update({"Age":24})
print(student)
```

```python
student["Age"] = 22                 # another way of update without method
print(student)
```

#### 6.pop()
###### Removes a specified key.

```python
student.pop("Age")
```

#### 7.popitem()
###### Removes the last inserted item.

```python
student.popitem()
print(student)
```

#### 8.clear()
###### Removes all items.

#### 9.copy()
###### Returns a copy of the dictionary.
