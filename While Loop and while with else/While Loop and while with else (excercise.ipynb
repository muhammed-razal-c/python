## While Loop and while with else

### Part A.

```python
# 1.
day = 1
while day <= 7:
    print("Day", day)
    day += 1
```

```python
# 2.
sales = [1200, 1500, 900, 2000, 1800]

i = 0
while i < 5:
    print(sales[i])
    i += 1
```

```python
# 3.
sales = [1200, 1500, 900, 2000, 1800]
total = 0
i = 0
while i < 5:
    total = total + sales[i]
    i += 1

print("Total sales:", total)
```

```python
# 4.
marks = [45, 78, 62, 90, 55]

total = 0
i = 0
while i < 5:
    total = total + marks[i]
    i += 1

average = total / 5
print("Average marks:", average)
```

```python
# 5.
marks = [45, 78, 62, 90, 55]

count = 0
i = 0
while i < 5:
    if marks[i] >= 50:
        count += 1
    i += 1

print("Students passed:", count)
```

```python
# 6.
quantity = int(input("Enter quantity: "))

i = 1
while i <= quantity:
    print(25 * i)
    i += 1
```

```python
# 7.
temperatures = [31, 34, 29, 36, 33]

highest = temperatures[0]
i = 1
while i < 5:
    if temperatures[i] > highest:
        highest = temperatures[i]
    i += 1

print("Highest temperature:", highest)
```

### Part B.

```python
# 1.
month = 1
while month <= 3:
    print(month)
    month += 1
else:
    print("Report Generated")
```

```python
# 2.
ids = [101, 102, 103, 104]

i = 0
while i < 4:
    print(ids[i])
    i += 1
else:
    print("All customers listed")
```
