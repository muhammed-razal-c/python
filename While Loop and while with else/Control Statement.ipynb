## 1.while loop

```python
#Question 1: Write a Python program to print numbers from 1 to 5.
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
#Question 2: Write a Python program to print numbers from 5 to 1
i = 5

while i >= 1:
    print(i)
    i -= 1
else :
    print("loop finished")
```

```python
#Question 3: Write a Python program to print even numbers from 2 to 10.
i = 2

while i <= 10:
    print(i)
    i += 2
else :
    print("loop finished")
```

```python
#Question 4: Write a Python program to print the multiplication table of 2.
i = 1

while i <= 10:
    print(2 * i)
    i += 1
else :
    print("loop finished")
```

```python
#Q: Write a Python program to find the sum of the first 10 natural numbers.
n=10
i=0
total=0

while i<=n:
    total=total+i
    i+=1
    print("Total sum :",total)
else :
    print("loop finished")

```

## 2.While Loop with else

#### The else block runs after the while loop finishes normally.
### Syntax
##### while condition:
#####    statements
##### else:
#####    statements


```python
i = 1

while i <= 3:
    print(i)
    i += 1
else:
    print("Loop Finished")
```

```python
#Question: A shop's sales for 5 days are stored in a list. Write a Python program to find the total and average sales.
sales = [1200,1500,900,2000,1800]

total = 0 
i = 0
while i<len(sales):
    total+=sales[i]
    i+=1
average = total/len(sales)

print("total sales :",total)
print("average sales :",average)
```
