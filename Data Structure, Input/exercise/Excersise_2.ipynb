```python
# D2
marks = [72, 85, 90, 65, 88, 76, 95, 58]
marks.sort(reverse=True)
print(marks)
print("top 3",marks[:3])
print("Bottom 3",marks[-3:])
print("Total:",sum(marks))
Average = sum(marks)/len(marks)
print("Average:",Average)
```

```python
# D3
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 9000, 18000, 21000, 17000]
print("Total:",sum(sales))
Average = round(sum(sales)/len(sales),2)        #Average function= Total sales by Total length
print("Average:",Average)
month = months[sales.index(max(sales))]         # to find the max sale month using index numbers of the sale
print("Best month:",month,"(",max(sales),")") 
count = 0
for i in sales:
    if i >= Average:
        count+=1
print("Months above Average:",count)        
```
